```python
import tkinter as tk

fenster = tk.Tk()
fenster.title("Alter eingeben")
fenster.geometry("400x300")

anzeige = tk.Label(fenster, text="Alter:", font=("Arial", 20), fg="black")
anzeige.pack(pady=20)

eingabe = tk.Entry(fenster, font=("Arial", 16))
eingabe.pack(pady=10)

def show_alter():
    alter = eingabe.get()
    anzeige.config(text=f"Alter: {alter}")

def hide_alter():
    anzeige.config(text="Alter: ****")

show_button = tk.Button(fenster, text="Show alter", command=show_alter)
show_button.pack(pady=10)

hide_button = tk.Button(fenster, text="Hide alter", command=hide_alter)
hide_button.pack(pady=10)

fenster.mainloop()
```
